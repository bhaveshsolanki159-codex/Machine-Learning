### How to automate a chrome

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
import time
# import undetected_chromedriver as uc
```

```python
s = Service("C:/Users/bhave/OneDrive/Desktop/chromedriver-win64/chromedriver.exe")
driver = webdriver.Chrome(service = s) # open chrome 

driver.get("https://google.com") # open url 
time.sleep(2) # if internet is slow and to let your url load properly take delay
```

```python
# Fetch the search input box using xpath
from selenium.webdriver.common.by import By

user_input = driver.find_element(by=By.XPATH, value='/html/body/div[2]/div[6]/form/div[1]/div/div[1]/div[2]/div[2]/textarea') # find search bar
user_input.send_keys('CampusX') # write campus x on search bar
time.sleep(2)

from selenium.webdriver.common.keys import Keys # used to import keys like enter-key etc
user_input.send_keys(Keys.ENTER) # click enter
time.sleep(15)

link = driver.find_element(by=By.XPATH, value='//*[@id="rso"]/div[1]/div/div/div/div/div/div/div/div[1]/div/span/a') # find url of CampusX and copy XPATH
link.click()
time.sleep(8)
```
