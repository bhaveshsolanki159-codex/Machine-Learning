```python
with open('smartprix.html','r',encoding='utf-8') as f:
    html = f.read()
```

```python
from bs4 import BeautifulSoup
import numpy as np
```

```python
soup = BeautifulSoup(html,'lxml')
containers = soup.find_all('div',{'class':'sm-product has-tag has-features has-actions'})

name = []
price = []
specs = []

sim = []
processor = []
ram_storage = []
battery = []
display = []
camera = []
memory_card = []
android = []

for i in containers:

    # Model 
    try:
        name.append(i.find('h2').text)
    except:
        name.append(np.nan)

    # Price
    try:
        price.append(i.find('span',{'class':'price'}).text)
    except:
        price.append(np.nan)

    # Rating
    try:
        specs.append(i.find('div', {'class': f'score rank-1-bg'}).find('b').text)
    except:
        try:
            specs.append(i.find('div', {'class': f'score rank-2-bg'}).find('b').text)
        except:
            try:
                specs.append(i.find('div', {'class': f'score rank-3-bg'}).find('b').text)
            except:
                specs.append(np.nan)

    # # features
    features = []
    try:
        feat = i.find('ul', {'class': 'sm-feat specs'}).find_all('li')


        for f in feat:
            features.append(f.get_text(strip=True))


    except:
        features = [np.nan]
    

    # Store in different lists
    # Store each feature
    sim.append(features[0] if len(features) > 0 else np.nan)
    processor.append(features[1] if len(features) > 1 else np.nan)
    ram_storage.append(features[2] if len(features) > 2 else np.nan)
    battery.append(features[3] if len(features) > 3 else np.nan)
    display.append(features[4] if len(features) > 4 else np.nan)
    camera.append(features[5] if len(features) > 5 else np.nan)
    memory_card.append(features[6] if len(features) > 6 else np.nan)
    android.append(features[7] if len(features) > 7 else np.nan)
```
