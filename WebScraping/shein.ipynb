### How to scrap a data from website having infinte scrolling

```python
# Import required modules
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
import time

# Specify the path of the ChromeDriver executable
s = Service("C:/Users/bhave/OneDrive/Desktop/chromedriver-win64/chromedriver.exe")

# Launch a new Chrome browser instance
driver = webdriver.Chrome(service=s)

# Open the target webpage
driver.get("https://www.sheinindia.in/s/jeans-209077")

# Wait for the page to load completely
time.sleep(2)

# -------------------------------------------------------------------
# Get the total height of the webpage using JavaScript.
# document.body.scrollHeight returns the height of the entire page,
# including the part that is not currently visible.
# -------------------------------------------------------------------
old_height = driver.execute_script("return document.body.scrollHeight")

# Scroll to the bottom of the page.
# This triggers lazy loading (loading more products as you scroll).
driver.execute_script("window.scrollTo(0, document.body.scrollHeight)")

# Wait for new content to load
time.sleep(5)

# Counter to track how many times the page has been scrolled
count = 1

# -------------------------------------------------------------------
# Keep scrolling until no new content is loaded.
# Logic:
# 1. Scroll to the bottom.
# 2. Wait for new products to load.
# 3. Measure the page height again.
# 4. If the height has not changed, we've reached the end.
# -------------------------------------------------------------------
while True:

    # Scroll to the bottom of the webpage
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight)")

    # Give the website time to load additional products
    time.sleep(2)

    # Get the updated page height
    new_height = driver.execute_script("return document.body.scrollHeight")

    # Print iteration number
    print(f"Scroll Count : {count}")
    count += 1

    # Print old and new heights for debugging
    print(f"Old Height : {old_height}")
    print(f"New Height : {new_height}")

    # If the page height hasn't increased,
    # no new content has been loaded.
    if new_height == old_height:
        print("Reached the end of the page.")
        break

    # Update old_height for the next comparison
    old_height = new_height

# -------------------------------------------------------------------
# Once all products are loaded, get the complete HTML source.
# Since we've scrolled through the entire page, the HTML now contains
# all dynamically loaded products.
# -------------------------------------------------------------------
html = driver.page_source

# Save the HTML into a file for later parsing
with open("Shein.html", "w", encoding="utf-8") as f:
    f.write(html)

print("HTML saved successfully.")
```
