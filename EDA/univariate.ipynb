### Univeriate Data Analysis

```python
# Opening a CSV file from URL

import requests
from io import StringIO
import pandas as pd
import seaborn as sns 

url = "https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/refs/heads/main/day20-univariate-analysis/train.csv"
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
req = requests.get(url, headers=headers)
data = StringIO(req.text)

df = pd.read_csv(data)
df.sample(5)
```

##### Categorical Data

```python
# how many people survied???

sns.countplot(x=df['Survived'],palette='Set2') # -> Give Frequency count of each unique value and most apply for categorial data
# alternative df['Survived'].value_counts().plot(kind = 'bar')
```

```python
# In which class most people are travelled??

sns.countplot(x=df['Pclass'],palette='pastel')
```

```python
# why people travelling 3 most bcz its fare is less as compared to 1 and 2.
df.groupby('Pclass')['Fare'].mean()
```

```python
# Boarding point Where most of people Board to ship is?

sns.countplot(x = df['Embarked'], palette='pastel')
```
