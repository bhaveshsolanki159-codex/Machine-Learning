```python
# Correlation tells us how two columns move together.
# Value ranges from -1 to 1:
#   +1  -> Both columns increase together (e.g. more votes = more popular)
#   -1  -> One increases while other decreases (inverse relation)
#    0  -> No relation at all
# Rule of thumb:
#   0.0 - 0.3  -> Weak correlation
#   0.3 - 0.7  -> Moderate correlation
#   0.7 - 1.0  -> Strong correlation

# ex-
# If I see column A go up, can I predict whether column B will go up, go down, or do nothing?
# Imagine if movies with more votes became less popular — as one goes up, the other goes down. That's "moving opposite.
# If I see column A go up, can I predict whether column B will go up, go down, or do nothing?

df.corr(numeric_only=True)['popularity']
```
